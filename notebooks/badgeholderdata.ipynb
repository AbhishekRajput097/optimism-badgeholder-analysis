### All Attestations from addresses "0x62...D7D" and "0xE4...226F"

```python
import requests
import csv
import json
from datetime import datetime

# Define the API endpoint and the query
url = 'https://optimism.easscan.org/graphql/playground'
query = '''
query Attestations($where: AttestationWhereInput) {
  attestations(where: $where) {
    attester
    data
    decodedDataJson
    expirationTime
    id
    ipfsHash
    isOffchain
    recipient
    refUID
    revocable
    revocationTime
    revoked
    schema {
      schemaNames {
        name
      }
    }
    schemaId
    time
    timeCreated
    txid
  }
}
'''

# Define the variables
variables = {
  "where": {
    # "attester": {
    #   "in": ["0x621477dBA416E12df7FF0d48E14c4D20DC85D7D9", "0xE4553b743E74dA3424Ac51f8C1E586fd43aE226F"]
    # },
    "schemaId": {
      "equals": "0x41513aa7b99bfea09d389c74aacedaeb13c28fb748569e9e2400109cbe284ee5"
    }
  }
}

# Function to convert Unix timestamp to date format
def format_timestamp(timestamp, with_time=False):
    dt_object = datetime.fromtimestamp(timestamp)
    if with_time:
        return dt_object.strftime('%m/%d/%Y %I:%M:%S %p')
    else:
        return dt_object.strftime('%m/%d/%Y')

# Send the request
response = requests.post(
    url,
    json={'query': query, 'variables': variables}
)

# Check if the request was successful
if response.status_code == 200:
    data = response.json()
    if 'data' in data and 'attestations' in data['data']:
        # Extract the attestations
        attestations = data['data']['attestations']

        # Define the CSV file name
        csv_file = 'attestations.csv'

        # Write to the CSV file
        with open(csv_file, 'w', newline='') as file:
            writer = csv.writer(file)
            # Write the header
            writer.writerow([
                'attester', 'rpgfRound', 'referredBy', 'referredMethod', 'expirationTime', 'id', 
                'ipfsHash', 'isOffchain', 'recipient', 'refUID', 'revocable', 'revocationTime', 
                'revoked', 'schemaNames', 'schemaId', 'time', 'timeCreated', 'txid'
            ])
            # Write the rows
            for attestation in attestations:
                decoded_data = json.loads(attestation['decodedDataJson'])
                
                # Extract values for the new columns
                rpgf_round = next((item['value']['value'] for item in decoded_data if item['name'] == 'rpgfRound'), '')
                referred_by = next((item['value']['value'] for item in decoded_data if item['name'] == 'referredBy'), '')
                referred_method = next((item['value']['value'] for item in decoded_data if item['name'] == 'referredMethod'), '')

                # Format time and timeCreated
                formatted_time = format_timestamp(attestation['time'])
                formatted_time_created = format_timestamp(attestation['timeCreated'], with_time=True)

                writer.writerow([
                    attestation['attester'], rpgf_round, referred_by, referred_method, 
                    attestation['expirationTime'], attestation['id'], attestation['ipfsHash'], 
                    attestation['isOffchain'], attestation['recipient'], attestation['refUID'], 
                    attestation['revocable'], attestation['revocationTime'], attestation['revoked'], 
                    ','.join([name['name'] for name in attestation['schema']['schemaNames']]), 
                    attestation['schemaId'], formatted_time, formatted_time_created, 
                    attestation['txid']
                ])

        print(f"Data has been written to {csv_file}")
else:
    print(f"Query failed with status code {response.status_code}")
    print(response.text)
```

### Data Sepration for different rounds

```python
import csv

# Open the CSV file
with open('attestations.csv', newline='') as csvfile:
    reader = csv.DictReader(csvfile)
    # Create empty lists to store rows for each rpgfRound value
    round_2_rows = []
    round_3_rows = []
    round_4_rows = []
    null_round_rows = []

    # Iterate over each row
    for row in reader:
        # Check the value of the rpgfRound column and append the row to the appropriate list
        if row['rpgfRound'] == '2':
            round_2_rows.append(row)
        elif row['rpgfRound'] == '3':
            round_3_rows.append(row)
        elif row['rpgfRound'] == '4':
            round_4_rows.append(row)
        elif not row['rpgfRound']:  # Check if rpgfRound value is null
            null_round_rows.append(row)

# Write rows to separate CSV files
def write_csv(filename, rows):
    with open(filename, 'w', newline='') as csvfile:
        writer = csv.DictWriter(csvfile, fieldnames=reader.fieldnames)
        writer.writeheader()
        for row in rows:
            writer.writerow(row)

# Write rows to separate CSV files based on rpgfRound value
write_csv('../data/rpgfround_2.csv', round_2_rows)
write_csv('../data/rpgfround_3.csv', round_3_rows)
write_csv('../data/rpgfround_4.csv', round_4_rows)
write_csv('../data/null_rpgfround.csv', null_round_rows)

print("Separation complete.")
```

```python
import pandas as pd

# Load the CSV files
rpgfRound_3 = pd.read_csv('../data/rpgfround_3.csv')
rpgfRound_4 = pd.read_csv('../data/rpgfround_4.csv')

# Extract the 'recipient' columns
recipients_3 = rpgfRound_3['recipient']
recipients_4 = rpgfRound_4['recipient']

# Find common addresses
common_addresses = recipients_3[recipients_3.isin(recipients_4)]

# Find addresses in rpgfRound_4 but not in rpgfRound_3
unique_addresses_4 = recipients_4[~recipients_4.isin(recipients_3)]

# Create a new DataFrame
result = pd.DataFrame({
    'common_addresses': pd.Series(common_addresses).reset_index(drop=True),
    'unique_addresses_4': pd.Series(unique_addresses_4).reset_index(drop=True)
})

# Save the new DataFrame to a CSV file
result.to_csv('../data/result.csv', index=False)
```
